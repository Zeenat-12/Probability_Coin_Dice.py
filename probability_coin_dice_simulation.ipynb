## 1. Import libraries
`random` generates random outcomes; `pandas` organizes tables; `matplotlib` plots.

```python
# Import Python's random module to make random choices
import random
# Import pandas for data tables
import pandas as pd
# Import matplotlib for drawing graphs
import matplotlib.pyplot as plt
```

## 2. One coin toss
A coin's sample space is `{Heads, Tails}`.

```python
# Randomly choose Heads or Tails for one toss
one_toss = random.choice(['Heads', 'Tails']) # builtin function: Return a random element from a list

print('One toss:', one_toss)
```

## 3. Simulate 1,000 coin tosses
The observed proportions may approach 50% as the number of trials grows, but are not guaranteed to be exactly 50%.

```python
# Set how many coin tosses to simulate
num_tosses = 1000
# Simulate each toss and collect results in a list
coin_results = [random.choice(['Heads', 'Tails']) for _ in range(num_tosses)]
# Count the number of Heads in the results
heads = coin_results.count('Heads')
# Count the number of Tails in the results
tails = coin_results.count('Tails')
# Calculate observed fraction of Heads
heads_rate = heads / num_tosses
# Calculate observed fraction of Tails
tails_rate = tails / num_tosses
# Display the counts and fractions
print('Tosses:', num_tosses, '| Heads:', heads, '| Heads proportion:', round(heads_rate, 3))
print('Tails:', tails, '| Tails proportion:', round(tails_rate, 3))
```

## 4. Plot coin results
The bars show the number of Heads and Tails observed.

```python
# Create labels for the coin outcomes
labels = ['Heads', 'Tails']
# Create matching bar heights from the counts
counts = [heads, tails]
# Draw the bar chart
plt.bar(labels, counts)
# Add a chart title
plt.title('Coin Toss Results (1,000 Tosses)')
# Label the horizontal axis
plt.xlabel('Outcome')
# Label the vertical axis
plt.ylabel('Count')
# Display the plot
plt.show()
```

## 5. Die sample space and event
A standard die sample space is `{1,2,3,4,5,6}`. The even-number event is `{2,4,6}`.

```python
# List all possible outcomes for a standard six-sided die
sample_space = [1, 2, 3, 4, 5, 6]
# Define the event of rolling an even number
even_event = [2, 4, 6]
# Display the sample space
print('Sample space:', sample_space)
# Display the even-number event
print('Event (even):', even_event)
# Calculate theoretical chance: favorable outcomes divided by all outcomes
theoretical_even = len(even_event) / len(sample_space)
# Print the theoretical probability
print('Theoretical probability of even:', theoretical_even)
```

## 6. Simulate 1,000 dice rolls
Randomly roll the die 1,000 times and count each face.

```python
# Choose how many rolls to simulate
num_rolls = 1000
# Simulate rolls by randomly selecting one face each time
dice_results = [random.choice(sample_space) for _ in range(num_rolls)]
# Count how often each face occurs
face_counts = {face: dice_results.count(face) for face in sample_space}
# Print the face counts
display(face_counts)
```

## 7. Plot dice results
Each face has theoretical probability 1/6 (about 16.7%), but random counts may vary.

```python
# Use die faces as the chart labels
faces = list(face_counts.keys())
# Use the observed counts as bar heights
frequencies = list(face_counts.values())
# Draw a bar chart
plt.bar(faces, frequencies)
# Add a title
plt.title('Dice Roll Results (1,000 Rolls)')
# Label horizontal axis
plt.xlabel('Die face')
# Label vertical axis
plt.ylabel('Count')
# Show chart
plt.show()
```

## 8. Estimate probability of an even roll
Compare experimental probability with the theoretical value 3/6 = 0.5.

```python
# Count results that are 2, 4, or 6
observed_even = sum(1 for value in dice_results if value in even_event)
# Divide even rolls by all rolls to estimate probability
experimental_even = observed_even / num_rolls
# Print the theoretical probability
print('Theoretical probability:', 3 / 6)
# Print the probability observed in the simulation
print('Experimental probability:', round(experimental_even, 3))
```

## 9. Experiment with trial counts
Change `num_tosses` or `num_rolls` to 10, 100, 1,000, or 10,000 and rerun the relevant cells. Results may fluctuate, but larger runs often settle closer to theoretical probabilities.

## 10. Recap
- **Random experiment:** an uncertain action (coin toss or die roll).
- **Sample space:** all possible outcomes.
- **Event:** the outcome(s) we're interested in.
- **Theoretical probability:** calculated from the mathematical model.
- **Experimental probability:** estimated from simulation results.
- Fair coin: Heads has probability 1/2.
- Fair die: even result has probability 3/6 = 1/2.
